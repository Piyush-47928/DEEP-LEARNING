```python
#IMPORTS
from ultralytics import YOLO
from google.colab import files
import cv2
import numpy as np
from collections import defaultdict
```

```python
#LOAD MODEL
model = YOLO('yolov8n.pt')
```

```python
#UPLOAD IMAGE
uploaded = files.upload()
image = list(uploaded.keys())[0]
```

```python
#READ IMAGE
img = cv2.imread(image)
h,w,_ = img.shape
```

```python
#RUN DETECTION
result = model(img)[0]
```

```python
#PROCESS RESULTS
object_counts = defaultdict(int)
detailed_results = []

for i, box in enumerate(result.boxes):
    cls_id = int(box.cls[0])
    conf = float(box.conf[0])
    label = model.names[cls_id]

    x1, y1, x2, y2 = box.xyxy[0].tolist()
    x1 /= w
    x2 /= w
    y1 /= h
    y2 /= h

    object_counts[label] += 1
    count_id = f"{label}_{object_counts[label]}"

    detailed_results.append({
        'Object': label,
        'Count ID': count_id,
        'Confidence': round(conf,4),
        'Bounding Box': (round(x1, 4), round(y1, 4),round(x2, 4) ,round(y2,4))
    })
```

```python
#PRINT STRUCTURED OUTPUT

total_objects = len(detailed_results)
print(f"\nTotal Objects Detected: {total_objects}\n")

print("- Objects Detected:")
for obj, count in object_counts.items():
    print(f"{obj}: {count}")

print("\n - Detailed Detection:")
for item in detailed_results:
    print(
        f'Object: {item["Object"]}\n',
        f'Count ID: {item["Count ID"]}\n',
        f'Confidence: {item["Confidence"]}\n',
        f'Bounding Box: {item["Bounding Box"]}\n'
    )
```

```python
#IMAGE WITH BOXES

annotated = result.plot()
annotated = cv2.cvtColor(annotated, cv2.COLOR_RGB2BGR)

import matplotlib.pyplot as plt
plt.imshow(annotated)
plt.show()
plt.axis('off')
```
